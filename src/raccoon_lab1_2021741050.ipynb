```python
import json
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path

# 경로 설정 및 에피소드 로드
root = Path("/data/2021741050/Raccoonbot_Openvla/Mujoco/raccoon_tasks_extended")
ep = sorted([p for p in root.iterdir() if p.is_dir() and p.name.startswith("episode_")])[0]

frames = sorted(ep.glob("frame_*.png"))
with open(ep / "meta.json", "r") as f:
    meta = json.load(f)

# 메타데이터 추출
instruction = meta.get("instruction", meta.get("language_instruction", "N/A"))
target_color = meta.get("target_color", "N/A")
success = meta.get("success", "N/A")

# [개선 1] 전체 에피소드의 흐름을 보기 위한 균등 샘플링 (Uniform Sampling)
n_show = min(5, len(frames))
# 0부터 마지막 프레임 인덱스까지 n_show 개수만큼 일정한 간격으로 추출
indices = np.linspace(0, len(frames) - 1, n_show, dtype=int)
sampled_frames = [frames[i] for i in indices]

# 시각화 설정
fig, axes = plt.subplots(1, n_show, figsize=(4 * n_show, 4))
if n_show == 1:
    axes = [axes] # 단일 프레임일 경우 iterable하게 리스트로 감싸기

for ax, idx, frame_path in zip(axes, indices, sampled_frames):
    img = Image.open(frame_path)
    ax.imshow(img)
    
    # [개선 2] 진행률(Step) 및 Action 상태 표기 (meta.json 구조에 맞춰 확장 가능)
    # 만약 meta 데이터에 프레임별 액션값(예: 그리퍼 개폐 여부)이 있다면 아래에 추가 가능합니다.
    step_info = f"Step: {idx}/{len(frames)-1}"
    
    ax.set_title(f"{frame_path.name}\n({step_info})", fontsize=10, pad=10)
    ax.axis("off")

# [개선 3] 타이틀 가독성 강화 및 VLA 정보 강조
plt.suptitle(
    f"🤖 [Episode Visualization] {ep.name}\n"
    f"Language Instruction: '{instruction}'\n"
    f"Target Color: {target_color} | Success: {success}",
    fontsize=14,
    fontweight="bold",
    y=1.05 # 타이틀이 이미지와 겹치지 않도록 여백 조정
)

plt.tight_layout()
plt.show()
```
